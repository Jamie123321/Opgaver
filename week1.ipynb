```python
# Import relevant libraries
import numpy as np
import sympy as sp

sp.init_printing()
```

Problem 306

```python
# Solve the equation x^4 + 2x^2 + x = 0
x = sp.Symbol('x')
equation = x**4 + 2*x**2 + 1

# Solve the equation
solutions = sp.solve(equation, x)

# Display the solutions simplified
[sp.N(s, 5) for s in solutions]
```

Problem 320

```python
# Compute (1 + i)^4
i = sp.I  # Define the imaginary unit
result = ((1+i) + 3) / ((1+i)**4 + 6*(1+i)**2 + 25)
result.simplify()
```
